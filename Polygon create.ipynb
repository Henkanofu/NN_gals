```python

image_size = (720,1280)

path = '/home/user/PycharmProjects/cv/example_lables/f2_178_jpg.rf.2034752a8f4e176b2bf862bcd336f723.txt'
```

```python
with open(path, 'r') as file:
    lines = file.readlines()

mask = np.zeros(image_size, dtype=np.uint8)
for line in lines:
    data = line.split("\n")[0].split(" ")[1:] # delete Mr. enter and whatever the classes it is
    data = list(map(float, data)) #make it list
    mask_points = np.array(data).reshape(-1, 2) #make it in xy,xy,xy,xy....
    coordinates = (mask_points * image_size[::-1]).astype(np.int32) #make it as 32image
    cv2.fillPoly(mask, [coordinates], (1))
```
